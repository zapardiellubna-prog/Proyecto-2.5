# Ejercicio 1: Sumatoria de no negativos (funcs.py)

```python
def sumatoria_positivos(nums):
    """Devuelve la suma de los valores mayores o iguales a 0 en una lista de enteros.

    :param nums: Lista de números enteros.
    :return: Suma total de los enteros no negativos.
    """
    total = 0
    for n in nums:
        if n >= 0:
            total += n
    return total
```

```python
def es_palindromo(s: str) -> bool:
    s = s.lower().replace(' ', '')
    return s == s[::-1]
```

# Ejercicio 3: Máximo con control de errores (funcs.py)

```python
def maximo_seguro(nums):
    """Devuelve el valor máximo de una lista no vacía de números.

    param nums: Lista de números (int o float).
    return: El valor numérico máximo.
    raises ValueError: Si la lista provista está vacía.
    """
    if not nums:
        raise ValueError('lista vacía')
    maximo = nums[0]
    for n in nums[1:]:
        if n > maximo:
            maximo = n
    return maximo
```

# Ejercicio 4: Contador Simple de palabras (funcs.py)

```python
def contar_palabras(texto: str) -> dict:
    """Cuenta la frecuencia de aparición de cada palabra en un texto.

    param texto: Cadena de texto a analizar.
    return: Diccionario con la relación palabra -> frecuencia.
    """
    conteo = {}
    for palabra in texto.lower().split():
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo
```

# Testing ejercicio 4: test_funcs.py
Texto vacío.
Palabras repetidas.
Comprobar que "Hola hola" devuelve {'hola': 2}.

```python
import pytest
def test_contar_palabras_texto_vacio():
    """Verifica que contar_palabras devuelva un diccionario vacío para un texto vacío."""
    assert contar_palabras("") == {}

def test_contar_palabras_insensible_a_mayusculas():
    """Verifica que contar_palabras trate las palabras de manera insensible a mayúsculas y minúsculas."""
    resultado = contar_palabras("Hola hola")
    assert resultado == {"hola": 2}

def test_contar_palabras_frase_compleja():
    """Verifica que contar_palabras cuente correctamente las palabras en una frase compleja."""
    texto = "Python es genial y probar código en Python es divertido"
    esperado = {
        "python": 2,
        "es": 2,
        "genial": 1,
        "y": 1,
        "probar":1,
        "código": 1,
        "en": 1,
        "divertido": 1,
    }
    assert contar_palabras(texto) == esperado
```

# Ejercicio 5: Filtrar Aprobados (funcs.py)

```python
def filtrar_aprobados(pares):
    """Filtra y devuelve los nombres de los estudiantes con una nota superior o igual a 5.

    param pares: Lista de tuplas compuestas por (nombre, nota).
    return: Lista de nombres de los estudiantes aprobados manteniendo el orden original.
    """
    resultado = []
    for nombre, nota in pares:
        if nota >= 5:
            resultado.append(nombre)
    return resultado
```

# Testing Ejercicio 5: test_funcs.py
Parametrizar 4 casos distintos.
Verificar que se mantiene el orden original.

```python
import pytest
@pytest.mark.parametrize("pares, esperado", 
    [
        ([("Alice", 6.0), ("Bob", 4.0), ("Charlie", 7.0)], ["Alice", "Charlie"]),
        ([("David", 5.0), ("Eve", 5.0)], ["David", "Eve"]),
        ([("Frank", 4.0), ("Grace", 3.0)], []),
        ([], []),
    ],
    ids= ["lista_mixta", "todos_aprobados", "ninguno_aprobado", "lista_vacia"]
)
def test_filtrar_aprobados(pares, esperado):
    assert filtrar_aprobados(pares) == esperado

def test_filtrar_aprobados_orden():
    """Verifica que filtrar_aprobados mantenga el orden original"""
    pares = [("Alice", 6.0), ("Bob", 4.0), ("Charlie", 7.0), ("David", 5.0)]
    esperado = ["Alice", "Charlie", "David"]
    assert filtrar_aprobados(pares) == esperado
```

# Ejercicio 6: Normalizar Email (funcs.py)

```python
def normalizar_email(email: str):
    """Separa y normaliza a minúsculas el usuario y el dominio de una dirección de e-mail.

    param email: Cadena con la dirección de correo electrónico.
    return: Tupla con (usuario, dominio) en minúsculas.
    raises ValueError: Si la dirección no contiene exactamente un símbolo '@'.
    """
    if email.count('@') != 1:
        raise ValueError('email inválido')
    usuario, dominio = email.split('@')
    return usuario.lower(), dominio.lower()
```

# Testing Ejercicio 6: test_funcs.py
Casos válidos e inválidos.
Comprobar normalización a minúsculas.

```python
import pytest
@pytest.mark.parametrize(
    "email, resultado_esperado",
    [
        ("alice@ejemplo.com", ("alice", "ejemplo.com")),
        ("BOB@EJEMPLO.COM", ("bob", "ejemplo.com")),
        ("charlie@subdominio.EJEMPLO.COM", ("charlie", "subdominio.ejemplo.com")),
        ("Alice@ejemplo.com", ("alice", "ejemplo.com"))
    ],
    ids=["email_normal", "email_con_mayusculas", "email_con_dominio_subdominio", "capitalizacion_mixta"]
)
def test_normalizar_email(email, resultado_esperado):
    """Comprueba la normalización a minúsculas en correos electrónicos válidos."""
    assert normalizar_email(email) == resultado_esperado
@pytest.mark.parametrize(
    "email_invalido",
    [
        "sin_arroba.com",
        "usuario@@dominio.com",
        "usuario@dominio@otro.com",
        "",
    ],
    ids=[
        "falta_arroba",
        "arroba_doble_consecutiva",
        "multiples_arrobas_separadas",
        "cadena_vacia",
    ],
)
def test_normalizar_email_invalido(email_invalido):
    """Valida la generación de ValueError cuando la estructura del email no es adecuada."""
    with pytest.raises(ValueError, match="email inválido"):
        normalizar_email(email_invalido)
```

# Ejercicio 7: Factorial iterativo con control de errores (funcs.py)

```python
def factorial(n: int) -> int:
    """Calcula el factorial de un número entero no negativo de forma iterativa.

    param n: Número entero mayor o igual a 0.
    return: Resultado del factorial de n.
    raises ValueError: Si n es un entero negativo.
    """
    if n < 0:
        raise ValueError('n negativo')
    resultado = 1
    for i in range(2, n+1):
        resultado *= i
    return resultado
```

# Testing Ejercicio 7: test_funcs.py
Casos (0,1), (1,1), (5,120).
Comprobar excepción para valores negativos.

```python
import pytest
@pytest.mark.parametrize(
    "num, resultado_esperado",
    [
        (0,1),
        (1,1),
        (5,120)
    ],
    ids=[
        "factorial_0",
        "factorial_1",
        "factorial_5"
    ],
)
def test_factorial_valido(num, resultado_esperado):
    """ Verifica el cálculo correcto del factorial para valores válidos."""
    assert factorial(num) == resultado_esperado
def test_factorial_negativo():
    """ Verifica que la función factorial lance ValueError al recibir un número negativo. """
    with pytest.raises(ValueError, match="n negativo"):
        factorial(-1)
```

# Ejercicio 8: Media aritmética (funcs.py)

```python
def media(nums):
    """Calcula la media aritmética de una lista no vacía de números.

    :param nums: Lista de números (int o float).
    :return: Valor flotante con el promedio aritmético.
    :raises ValueError: Si la lista está vacía.
    """
    if not nums:
        raise ValueError('lista vacía')
    return sum(nums) / len(nums)
```
